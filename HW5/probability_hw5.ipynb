## Bayes Network

```python
network = BayesNet([
    ('Smoker', '', (0.3)),
    ('Cancer', 'Smoker', {(True, ): 0.05, (False, ): 0.01}),
    ('X-ray', 'Cancer', {(True, ): 0.9, (False, ): 0.2})
])
```

```python
smoker_node = network.variable_node('Smoker')
cancer_node = network.variable_node('Cancer')
xray_node = network.variable_node('X-ray')
```
